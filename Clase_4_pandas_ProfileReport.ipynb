```python
#Graficos
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,10)) #Creamos la figura y fijamos el tamaño
plt.hist(df['CREDIT'], bins = 10) #Tomamos las divisiones
```

```python
#Lo que hace el histograma basicamente es escoger un rango y realizar una distribuión de frecuencias =
#con los datos que se le estan proporcionando.

#Es curioso porque no es una grafica comporativa de x vs y sino es una grafica que muestra como las frecuencias
# de los datos estan distribuidas en el dataset.
plt.figure(figsize=(10,10))
plt.hist(df['CREDIT'], bins = 1000 , histtype='stepfilled', align = 'right', color = 'orange', orientation='horizontal')
```

```python
#Vamos a hacerlo con seaborn

import seaborn as sns
```

```python
sns.histplot(df['CREDIT'])
```

```python
#Otros ejemplos para categorias

sns.set_theme(style="ticks", palette="pastel")

# Example: Boxplot of 'CREDIT' based on 'CODE_GENDER'
sns.boxplot(x="CODE_GENDER", y="CREDIT",
            hue="CODE_GENDER", # Add hue to address FutureWarning
            palette=["m", "g", "b"], # Provide 3 colors for 'F', 'M', 'XNA' to address UserWarning
            data=df,
            legend=False) # Set legend=False as suggested by FutureWarning
sns.despine(offset=10, trim=True)
plt.title('Credit Amount by Gender')
plt.xlabel('Gender')
plt.ylabel('Credit Amount')
plt.show()
```
